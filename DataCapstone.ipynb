```python
#which country has the highest population
df[df['Population'] == df['Population'].max()]['Country']
```

```python
#What is the Area (sq. mi.) of the highest population country
df[df['Population'] == df['Population'].max()]['Area (sq. mi.)']
```

```python
#which country has the least population
df[df['Population'] == df['Population'].min()]['Country']
```

```python
#What is the Area (sq. mi.) of the least population country
df[df['Population'] == df['Population'].min()]['Area (sq. mi.)']
```

```python
#Top 5 countries with highest GDP ($ per capita) 
df.nlargest(5, "GDP ($ per capita)")
```

```python
#How many total regions are there
df['Region'].value_counts().count()
```

```python
#How many countries lie in Eastern Europe region
df['Region'] = df['Region'].str.strip()
df['Region'].value_counts()['EASTERN EUROPE']
```

```python
#What is the Region of the second highest population country
df[df['Population'] == df['Population'].nlargest(2).iloc[1]]['Region']
```

```python
#How many countries are there whose GDP ($ per capita) is unknown
df[df['GDP ($ per capita)'].isna()]['Country'].count()
```

```python
#Which country in African region has the highest population
africa_df = df[df['Region'].str.contains('AFRICA')]
```

```python
africa_df[africa_df['Population'] == africa_df['Population'].max()]['Country']
```

```python
#how many country has republic in their full name
count = 0
def counting(text) : 
    global count
    if 'rep' in text.lower() : 
        count += 1
    return text

df['Country'] = df['Country'].apply(counting)
print(count)
```
