## DataFrames :

```python
import numpy as np
import pandas as pd
```

Creating a DataFrames :

```python
data = {
    'Name' : ['John', 'Anna', 'Peter', 'Linda'],
    'Age' : [28, 34, 29, 42],
    'City' : ["Jaipur", 'Paris', 'Delhi', 'Mumbai'],
    'Salary' : [65000, 70000, 62000, 85000]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
data_list = [
    ['John', 28, "Jaipur", 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Delhi', 62000 ],
    ['Linda', 42 , 'Mumbai', 85000]
]
df2 = pd.DataFrame(data_list)
```

```python
df2
```

```python
col = ['Name', 'Age', 'City', 'Salary']
df2 = pd.DataFrame(data_list, columns = col)
df2
```

Selection and Indexing of Columns :

```python
df2['Name']
```

```python
df2[['Name','City']]
```

Creating a New Column :

```python
df2['Designation'] = ['Doctor', 'Engineer', 'Doctor', 'Engineer']
```

```python
df2
```

Remove a Column :

```python
df2.drop('Designation',axis = 1)
```

```python
df2
```

```python
df2.drop('Designation',axis = 1,inplace = True)
```

```python
df2
```

```python
df2.drop(0,axis = 0)
```

Selecting Rows :

```python
df2.loc[0]
```

```python
df2.loc[[0,1]]
```

Selecting Subsets of Rows and Columns :

```python
df.loc[[0,1]][['City','Salary']]
```

```python
df2.loc[[2,3]][['Name','Age']]
```

Conditional Selection :

```python
#People whose age is 30 and above

df2[df2['Age'] > 30]
```

```python
#People whose age is 30 and above and their city must be Paris

df2[(df2['Age'] > 30) & (df2['City'] == 'Paris')]
```
